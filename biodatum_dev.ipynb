```python
# BIODATUM – Development Notebook

Climate and Bioeconomic Intelligence for the Amazon
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
years = np.arange(2000, 2023)
temperature = 26 + 0.02*(years - 2000)

plt.plot(years, temperature)
plt.title("Temperature Trend")
plt.show()
income = 100 - (temperature - 26)*12

plt.plot(years, income)
plt.title("Economic Impact")
plt.show()
biomass = 320 - (temperature - 26)*20

plt.plot(years, biomass)
plt.title("Forest Biomass")
plt.show()
## Insights

- Temperature increase leads to biomass loss
- Economic decline follows ecological degradation
- Strong causal relationship between climate and bioeconomy
```
