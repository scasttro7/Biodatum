## 📗 NÍVEL 2 — NumPy
### Cálculos científicos rápidos

**NumPy** é a biblioteca de matemática do Python. Ela faz cálculos em listas inteiras de uma vez — como aplicar uma fórmula pra uma coluna inteira de planilha.

````python
import numpy as np  # 'np' é o apelido padrão

# Transformando nossa lista em array NumPy
daps = np.array([85, 120, 45, 60, 150, 55, 180, 30, 70, 25])

# Agora podemos fazer contas em todas de uma vez!
print("DAPs:", daps)
print("Média:",  np.mean(daps).round(1))
print("Mínimo:", np.min(daps))
print("Máximo:", np.max(daps))
print("Desvio padrão:", np.std(daps).round(2))

# log natural — é o que usamos na equação de Higuchi!
ln_daps = np.log(daps)
print("\nln(DAPs):", ln_daps.round(3))
````

### 🎯 Exercício 4 — A equação de Higuchi na mão

Agora você vai calcular a biomassa de UMA árvore passo a passo.

A equação é:
```
AGB = exp(-1.499 + 2.148*ln(DAP) + 0.207*ln(DAP)² - 0.0281*ln(DAP)³)
```

Use `dap = 100` e preencha cada etapa abaixo:

````python
dap = 100

# Passo 1: calcule o ln do DAP
ln_dap = np.log(dap)
print(f"ln({dap}) = {ln_dap:.4f}")

# Passo 2: monte o expoente (complete os ???)
expoente = -1.499 + 2.148 * ln_dap + 0.207 * ln_dap**2 - 0.0281 * ln_dap**3
print(f"Expoente = {expoente:.4f}")

# Passo 3: calcule a AGB
agb = np.exp(expoente)
print(f"\n🌳 Biomassa da árvore com DAP={dap}cm: {agb:.2f} Mg/ha")

# 🎯 AGORA É COM VOCÊ:
# Mude o dap para 50, 150 e 200 e veja como a biomassa muda
# O que você percebe?
````

## 📙 NÍVEL 3 — Pandas
### Trabalhando com tabelas

**Pandas** é o Excel do Python. Você cria tabelas (chamadas de `DataFrame`) e faz filtros, cálculos e agrupamentos.

````python
import pandas as pd

# Criando um DataFrame — é como montar uma planilha no código
parcela = pd.DataFrame({
    'especie':  ['Bertholletia excelsa', 'Dinizia excelsa', 'Ceiba pentandra',
                 'Swietenia macrophylla', 'Manilkara huberi'],
    'dap_cm':   [85, 180, 150, 60, 45],
    'altura_m': [35, 50, 45, 28, 25]
})

# Ver a tabela
print(parcela)
print("\nFormato da tabela:", parcela.shape, "(linhas, colunas)")
````

````python
# Adicionar uma coluna nova com cálculo
parcela['agb'] = np.exp(
    -1.499
    + 2.148  * np.log(parcela['dap_cm'])
    + 0.207  * np.log(parcela['dap_cm'])**2
    - 0.0281 * np.log(parcela['dap_cm'])**3
).round(2)

print(parcela[['especie', 'dap_cm', 'agb']])
print(f"\n📊 Biomassa total da parcela: {parcela['agb'].sum():.2f} Mg/ha")
````

````python
# Filtrar apenas árvores grandes (DAP > 100)
grandes = parcela[parcela['dap_cm'] > 100]
print("Árvores com DAP > 100 cm:")
print(grandes[['especie', 'dap_cm', 'agb']])

# Ordenar por biomassa
print("\nOrdenado por biomassa:")
print(parcela.sort_values('agb', ascending=False)[['especie', 'agb']])
````

## 🔴 NÍVEL 5 — IRFA na prática
### Mexendo nos pesos e cenários

````python
# Função IRFA completa
def calcular_irfa(tnr, trb, iec, alpha=0.33, beta=0.33, gamma=0.33):
    """
    IRFA = α*TNR + β*TRB + γ*IEC
    
    TNR = Taxa de Regeneração Natural
    TRB = Taxa de Recuperação de Biomassa  
    IEC = Índice de Estrutura Comunitária
    """
    return alpha * tnr + beta * trb + gamma * iec

def classificar(irfa):
    if irfa < 0.4:  return '🔴 Baixa'
    elif irfa <= 0.7: return '🟡 Média'
    else:           return '🟢 Alta'

# Calculando IRFA para diferentes situações reais
situacoes = [
    ('Reserva Ducke (Manaus)',      0.88, 0.85, 0.90),
    ('Área após manejo sustentável',0.55, 0.50, 0.58),
    ('Margem de estrada AM-010',    0.20, 0.15, 0.18),
    ('Área em regeneração (5 anos)',0.38, 0.42, 0.35),
]

print(f"{'Local':<35} {'IRFA':>6}  {'Classificação'}")
print('-' * 60)
for nome, tnr, trb, iec in situacoes:
    irfa = calcular_irfa(tnr, trb, iec)
    print(f"{nome:<35} {irfa:>6.3f}  {classificar(irfa)}")
````

````python
# E SE os pesos fossem diferentes?
# Hipótese: regeneração natural é mais importante (peso maior)

print("Comparação de pesos — impacto no IRFA")
print(f"{'Local':<35} {'Pesos iguais':>14} {'TNR prioritário':>16}")
print('-' * 68)

for nome, tnr, trb, iec in situacoes:
    irfa_igual    = calcular_irfa(tnr, trb, iec, 0.33, 0.33, 0.33)
    irfa_tnr_prio = calcular_irfa(tnr, trb, iec, 0.60, 0.20, 0.20)  # TNR tem peso 60%
    print(f"{nome:<35} {irfa_igual:>14.3f} {irfa_tnr_prio:>16.3f}")

print("\n💡 Dica: mudar os pesos é uma decisão científica e metodológica.")
print("   No seu doutorado, você vai definir pesos baseados em literatura ou calibração de campo.")
````
